```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/diabetes_prediction_dataset.csv")

# First 5 rows
print(df.head())

# Dataset size
print("\nDataset Shape:")
print(df.shape)

# Dataset information
print("\nDataset Information:")
df.info()

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Statistics
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Count diabetes classes
print(df["diabetes"].value_counts())

# Calculate percentage
print("\nPercentage Distribution:")
print(df["diabetes"].value_counts(normalize=True) * 100)
```

The dataset contains 100,000 records and 9 attributes. The target variable, diabetes, is imbalanced, with 91.5% non-diabetic records and 8.5% diabetic records. Therefore, model performance will be evaluated using metrics such as precision, recall, F1-score, and ROC-AUC rather than accuracy alone.

```python
 # Check unique values in categorical columns

print("Gender:")
print(df["gender"].value_counts())

print("\nSmoking History:")
print(df["smoking_history"].value_counts())
```

The dataset contains two categorical features: gender and smoking history. Gender consists mainly of Female and Male categories, with a small number of Other records. Smoking history contains six categories, including No Info, which represents unknown smoking information. These categorical features will be encoded before model training.

```python
# Check duplicate rows
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Display some duplicate rows
df[df.duplicated()].head()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check the new dataset shape
print("New Dataset Shape:", df.shape)

# Confirm duplicates
print("Remaining Duplicates:", df.duplicated().sum())
```

The dataset initially contained 3,854 duplicate records. These duplicate rows were removed during preprocessing, resulting in 96,146 unique records for further analysis and model development.

The dataset was checked for duplicate records. A total of 3,854 duplicate rows were identified and removed, resulting in a cleaned dataset containing 96,146 unique records.

```python
# Check target distribution after removing duplicates

print(df["diabetes"].value_counts())

print("\nPercentage Distribution:")
print((df["diabetes"].value_counts(normalize=True) * 100).round(2))
```

After removing duplicate records, the target variable distribution was analyzed again. The cleaned dataset contains 87,664 non-diabetic records (91.18%) and 8,482 diabetic records (8.82%). The dataset therefore remains imbalanced, and evaluation metrics such as precision, recall, F1-score, and ROC-AUC will be considered during model evaluation.

```python
# Separate input features and target

X = df.drop("diabetes", axis=1)
y = df["diabetes"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

```python
# One-Hot Encode categorical columns

X_encoded = pd.get_dummies(
    X,
    columns=["gender", "smoking_history"],
    drop_first=True,
    dtype=int
)

print("Encoded Features Shape:", X_encoded.shape)

print("\nEncoded Feature Columns:")
print(X_encoded.columns.tolist())

print("\nFirst 5 Rows:")
print(X_encoded.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Target:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Define categorical and numerical columns
categorical_features = [
    "gender",
    "smoking_history"
]

numerical_features = [
    "age",
    "hypertension",
    "heart_disease",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level"
]

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ]
)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data
X_test_processed = preprocessor.transform(X_test)

print("Processed Training Shape:", X_train_processed.shape)
print("Processed Testing Shape:", X_test_processed.shape)
```

**MODEL TRAINING**

```python
from sklearn.linear_model import LogisticRegression

# Create the model
logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

# Train the model
logistic_model.fit(X_train_processed, y_train)

print("Logistic Regression model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Make predictions
y_pred = logistic_model.predict(X_test_processed)

# Get prediction probabilities
y_prob = logistic_model.predict_proba(X_test_processed)[:, 1]

# Metrics
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC Score:", roc_auc_score(y_test, y_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

The Logistic Regression model achieved an accuracy of 88.46% and an ROC-AUC score of 96.01%. The model achieved a recall of 87.97% for the diabetic class, correctly identifying 1,492 out of 1,696 diabetic cases. However, the precision was 42.53%, indicating a relatively high number of false-positive predictions. Therefore, additional machine learning models will be trained and compared before selecting the final model.

**RANDOM_FOREST**

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

# Train the model
rf_model.fit(X_train_processed, y_train)

print("Random Forest model trained successfully!")
```

```python
# Make predictions
y_pred_rf = rf_model.predict(X_test_processed)

# Get prediction probabilities
y_prob_rf = rf_model.predict_proba(X_test_processed)[:, 1]

# Metrics
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("F1 Score:", f1_score(y_test, y_pred_rf))
print("ROC-AUC Score:", roc_auc_score(y_test, y_prob_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
import joblib
import pandas as pd

# Load saved files
model = joblib.load("medintel_xgb_model.pkl")
preprocessor_loaded = joblib.load("medintel_preprocessor.pkl")
threshold = joblib.load("medintel_threshold.pkl")

# Create sample patient data
sample_patient = pd.DataFrame({
    "gender": ["Female"],
    "age": [50],
    "hypertension": [0],
    "heart_disease": [0],
    "smoking_history": ["never"],
    "bmi": [28.5],
    "HbA1c_level": [6.2],
    "blood_glucose_level": [150]
})

# Apply the same preprocessing
sample_processed = preprocessor_loaded.transform(sample_patient)

# Get diabetes probability
probability = model.predict_proba(sample_processed)[0][1]

# Apply our threshold
prediction = 1 if probability >= threshold else 0

print("Diabetes Risk Probability:", round(probability * 100, 2), "%")

if prediction == 1:
    print("Result: Higher Risk")
else:
    print("Result: Lower Risk")
```
